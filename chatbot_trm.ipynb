```python
from google import genai
import discord
from google.colab import userdata


GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=GEMINI_API_KEY)

chat = client.chats.create(
    model="gemini-3.6-flash",
    config={
        "system_instruction":""" Você é o Mestre dos Magos, um arcanista e especialista no sistema de RPG Tormenta20 e sabe tirar qualquer dúvida em relação as regras do sistema, de acordo com o que é escrito no próprio livro de regras do sistema. Responda de forma clara, concisa e de fácil entendimento, sempre indicando de onde aquela regra veio, conforme o livro de regras, quando apropriado. """
    }
)

DISCORD_BOT_TOKEN = userdata.get("DISCORD_BOT_TOKEN")

intents = discord.Intents.default()
intents.message_content = True

bot = discord.Client(intents=intents)

@bot.event
async def on_ready():
  print(f"Bot conectado como {bot.user}")

@bot.event
async def on_message(message):
  if message.author == bot.user:
    return

  if bot.user not in message.mentions:
    return

  content = message.content.replace(f"<{bot.user.id}>", "").strip()
  answer = chat.send_message(message.content)

  await message.channel.send(answer.text)

await bot.start(DISCORD_BOT_TOKEN)
```
